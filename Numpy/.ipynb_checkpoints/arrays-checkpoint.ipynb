
| Feature          | Python List               | NumPy Array    (NumPy)           |
| ---------------- | ------------------------- | -------------------------------- |
| Purpose          | General-purpose container | Numerical computing              |
| Data type        | Can store mixed types     | Homogeneous (same type only)     |
| Example          | `[1, "a", 3.5]`           | `[1, 2, 3]`                      |
| Performance      | Slow for large data       | Very fast (C-optimized)          |
| Operations       | Requires loops            | Vectorized (no loops needed)     |
| `+` operator     | Concatenation             | Element-wise addition            |
| Memory usage     | More (stores references)  | Less (compact storage)           |
| Multidimensional | Nested lists              | Built-in multi-dimensional       |
| Use case         | Basic Python programming  | Data science, ML, math-heavy ops |

Deep and Shallow Copy

```python
arr8 = np.arange(1,21)
arr8
```

```python
slice = arr8[:5]
```

```python
slice
```

```python
slice = slice * 10
```

```python
slice
```

```python
arr8
```

# In case of array, changes made in slice didnot reflect back to original array but in case it list chanegs will be reflected
# array me jab bhi slice krte hai to wo ek alag location/space bana lete hai so chanegs reflect nhi hote hai and this is shallow copy

# incase of deep copy chanegs will be reflected

```python
# If we are assigning a into any other variable then chanegs will be reflected
b = arr8
```

```python
b
```

```python
b[0] = 99
```

```python
b
```

```python
arr8
# see changes are reflected if we had made change in b
```

```python
# If we want ki b ke andar changes krne pe arr8 me reflect na ho then use .copy()

b = arr8.copy()
b
```

```python
b[1] = 128
b
```

```python
arr8
# index 1 remain unaffected 
```

# Advanced Array Multiplication

```python
# 1. Stacking arrays
arr9 = np.array([1,2,3,4])
arr10 = np.array([5,6,7,8])
```

```python
np.vstack((arr9,arr10))
#vertically stack
```

```python
np.hstack((arr9,arr10))
#horizontally stack
```

```python
# column stack
np.column_stack((arr9,arr10))
```
